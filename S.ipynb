```python
import numpy as np
import scipy.linalg as la
import matplotlib.pyplot as plt  # <-- ADDED MISSING IMPORT

N = 165

I_matrix = np.eye(N, dtype=complex)
# Assumes S_global is already defined in your environment
M = I_matrix - (C_matrix_perfect @ S_global) 

source_indices = {
    "Puttalam_Source": 0,
    "Barge_Source": 96,
    "Rantambe_Source": 150
}

# --- NEW: Dictionaries to store a, b, and V for each source ---
A_individual = {}
B_individual = {}
V_individual_complex = {}
V_individual_mag = {}

print("=== SOLVING FOR A, B, AND V FOR EACH SOURCE ===")

# ==========================================
# 2. SOLVE ONE SOURCE AT A TIME
# ==========================================
for src_name, k in source_indices.items():
    print(f"Calculating network response for {src_name} ONLY...")
    
    a_vec = np.zeros((N, 1), dtype=complex)
    a_vec[k, 0] = 1.0 + 0j  
    
    mask = np.arange(N) != k
    
    M_sub = M[np.ix_(mask, mask)]  
    M_col = M[mask, k].reshape(-1, 1)             
    
    # Solve for unknown incident waves
    rhs = -M_col * a_vec[k, 0]
    a_sub = la.solve(M_sub, rhs)
    
    # Reassemble 'a' and calculate 'b'
    a_vec[mask, 0] = a_sub.flatten()
    b_vec = S_global @ a_vec
    
    # --- STORE THE SEPARATE VALUES HERE ---
    A_individual[src_name] = a_vec.copy()
    B_individual[src_name] = b_vec.copy()
    V_individual_complex[src_name] = a_vec + b_vec
    
    # Calculate Worst-Case Voltage Magnitude as |a| + |b| 
    V_individual_mag[src_name] = np.abs(a_vec) + np.abs(b_vec)

# ==========================================
# 3. SUPERIMPOSE ALL RESULTS
# ==========================================
print("\nSuperimposing absolute voltage magnitudes...")

# V_total = |V_src1| + |V_src2| + |V_src3| (Absolute worst-case summation)
V_total_mag = sum(V_individual_mag.values())
a_total = sum(A_individual.values())
b_total = sum(B_individual.values())

nodes = np.arange(N)

plt.figure(figsize=(12, 6))

# Plot individual worst-case source contributions
plt.plot(nodes, V_individual_mag["Puttalam_Source"], label='Puttalam Only (|a|+|b|)', alpha=0.6, linestyle='--')
plt.plot(nodes, V_individual_mag["Barge_Source"], label='Barge Only (|a|+|b|)', alpha=0.6, linestyle='--')
plt.plot(nodes, V_individual_mag["Rantambe_Source"], label='Rantambe Only (|a|+|b|)', alpha=0.6, linestyle='--')

# Plot the Superimposed Total
plt.plot(nodes, V_total_mag, label='Worst-Case Total (|a|+|b| summed)', color='black', linewidth=2)

plt.title('Worst-Case Harmonic Voltage Envelope (250 Hz)', fontsize=14)
plt.xlabel('System Node Index', fontsize=12)
plt.ylabel('Voltage Magnitude (pu)', fontsize=12)
plt.grid(True, linestyle=':', alpha=0.7)
plt.legend(fontsize=11)
plt.tight_layout()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
from matplotlib import animation
from IPython.display import HTML

# 1. Define Junction Parameters (3-Port Junction)
n_ports = 3

# Generate the ideal loss-free S-matrix
S_matrix = np.full((n_ports, n_ports), 2.0 / n_ports, dtype=complex)
np.fill_diagonal(S_matrix, -(n_ports - 2) / n_ports)

# 2. Set up wave and spatial properties
x = np.linspace(10, 0, 300)  # Distance along the line (Junction is at x=0)
k = 2.0 * np.pi / 5.0        # Wave number
omega = 2.0 * np.pi / 2.0    # Angular frequency

# 3. Define the Source Contributions for each Port
# Port 1 has a strong source (1.0 pu)
# Port 2 has a weaker source (0.5 pu)
# Port 3 has no active source (0.0 pu), it only receives transmissions
A_inc = np.array([1.0, 0.5, 0.0])
phase_inc = np.array([0.0, np.pi/4, 0.0]) 

# 4. Initialize the plot
fig, axes = plt.subplots(n_ports, 1, figsize=(10, 8), sharex=True)
fig.suptitle('Wave Superposition at a 3-Port Junction\n(Multiple Source Contributions)', fontsize=14)

lines_inc = []
lines_ref = []
lines_tot = []

for i, ax in enumerate(axes):
    ax.set_xlim(10, 0)
    ax.set_ylim(-2.5, 2.5)
    ax.set_ylabel(f'Line {i+1} Volts')
    ax.grid(True, linestyle=':', alpha=0.7)
    
    # Initialize the lines for animation (using raw strings r'' for LaTeX)
    l_inc, = ax.plot([], [], 'b--', alpha=0.6, label=r'Incident Wave ($\vec{a}$)')
    l_ref, = ax.plot([], [], 'r--', alpha=0.6, label=r'Scattered Wave ($\vec{b}$)')
    l_tot, = ax.plot([], [], 'k-', linewidth=2, label='Total Standing Wave')
    
    lines_inc.append(l_inc)
    lines_ref.append(l_ref)
    lines_tot.append(l_tot)
    
    if i == 0:
        ax.legend(loc='upper right')

axes[-1].set_xlabel('Distance to Junction (Junction is at x=0)')

# 5. Animation functions
def init():
    for l_inc, l_ref, l_tot in zip(lines_inc, lines_ref, lines_tot):
        l_inc.set_data([], [])
        l_ref.set_data([], [])
        l_tot.set_data([], [])
    return lines_inc + lines_ref + lines_tot

def animate(frame):
    t = frame * 0.05
    
    # Calculate the complex values of all incident waves (a) at the junction (x=0) at this exact time
    a_complex = A_inc * np.exp(1j * (omega * t + phase_inc))
    
    # Calculate the resulting scattered waves (b) using the S-matrix
    # This automatically handles the superposition of all transmissions and reflections!
    b_complex = S_matrix.dot(a_complex)
    
    for i in range(n_ports):
        # Incident wave travels towards x=0
        V_inc = A_inc[i] * np.cos(omega * t - k * x + phase_inc[i])
        
        # Scattered wave (b) travels away from x=0
        A_b = np.abs(b_complex[i])
        phase_b = np.angle(b_complex[i])
        V_ref = A_b * np.cos(omega * t + k * x + phase_b)
        
        # The physical standing wave on the line is the sum of a and b
        V_tot = V_inc + V_ref
        
        lines_inc[i].set_data(x, V_inc)
        lines_ref[i].set_data(x, V_ref)
        lines_tot[i].set_data(x, V_tot)
        
    return lines_inc + lines_ref + lines_tot

# 6. Create and display the animation
ani = animation.FuncAnimation(
    fig, animate, frames=100, init_func=init, 
    interval=50, blit=True
)

plt.close(fig)
HTML(ani.to_jshtml())
```
