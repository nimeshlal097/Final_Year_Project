```python
import numpy as np

C=matrix
def perform_propagation_analysis(S, C, a1_known=1.0, source_idx=0):
    """
    Implements Equation 4.67: M(2:n,2:n) * a(2:n) = -M(2:n,1) * a1
    """
    n = S.shape[0]
    I = np.eye(n, dtype=complex)
    
    # Calculate Network Matrix M (Equation 4.66)
    # M = I - C*S
    M = I - np.matmul(C, S)
    
    # Indices for the unknown waves (indices 1 to 118 in Python logic)
    unknown_indices = np.delete(np.arange(n), source_idx)
    
    # --- Partitioning per Equation 4.67 ---
    
    # Left Hand Side: M(2:n, 2:n)
    A_matrix = M[np.ix_(unknown_indices, unknown_indices)]
    
    # Right Hand Side: -M(2:n, 1) * a1
    # Note: source_idx is usually 0 (Node 1)
    b_vector = -M[unknown_indices, source_idx] * a1_known
    
    # --- Solve the Linear System ---
    try:
        # Solve for the incident waves a2 through an
        a_unknown = np.linalg.solve(A_matrix, b_vector)
    except np.linalg.LinAlgError:
        # Fallback if matrix is singular (common in large ideal models)
        print("System is singular, using least-squares solver.")
        a_unknown = np.linalg.lstsq(A_matrix, b_vector, rcond=None)[0]
    
    # --- Final Results Construction ---
    
    # Incident Wave Vector (A)
    a_full = np.zeros(n, dtype=complex)
    a_full[source_idx] = a1_known
    a_full[unknown_indices] = a_unknown
    
    # Propagated Wave Vector (B) (Equation 4.60: b = S * a)
    b_full = np.matmul(S, a_full)
    return a_full, b_full

# --- Execution ---
# Replace S_data and C_data with your actual variables
a_results, b_results = perform_propagation_analysis(S, C, a1_known=1.0)

print("Incident Waves (A):")
print(a_results)
print("Reflected Waves (B):")
print(b_results)
```
